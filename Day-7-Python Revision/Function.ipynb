```python
def firstDigit(n):
    while n >= 10:
        n = n // 10
    else:
        return n


n = int(input())
print(firstDigit(n))
```

```python
def GFG(n):
    for i in range(n):
        print("GFG",end=" ")
n=int(input())
GFG(n)
```
