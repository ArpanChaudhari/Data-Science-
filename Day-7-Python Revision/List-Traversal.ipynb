```python
#  print the elements of the list with a space between them.
arr=[54,43,2,1,5]

# Method 1
for i in arr:
    print(i,end=" ")

print()
# Method 2
print(*arr,end=" ")
```

```python
# return the length of the list.
print(len(arr))
```

```python
# Return the sum of the list.
print(sum(arr))
# Return the max element of the list.
print(max(arr))
# Return the min element of the list.
print(min(arr))
```

```python
# decrement each element of the list by 1 and return the list.
for i in range(0,len(arr)):
    arr[i]-=1
        
print(arr)
```

```python
k = 6
for i in range(0, len(arr)):
    if arr[i] < k:
        print(arr[i], end=" ")

print()
# Method 2
ans = [i for i in arr if i < k]
print(ans)
```

```python
# Mean or Average of an list
print(sum(arr)/len(arr))
# Note: Return the floor value of the mean.
print(sum(arr)//len(arr))
```
