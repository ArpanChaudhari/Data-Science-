```python
# print the multiplication table from 1 to 10 for n in a single line, separated by spaces.
n=int(input())
for i in range(1,11):
    print(n*i,end=" ")
```

```python
# String S, you need to print its characters at even indices(index starts at 0).
def utility(s):

    # Method 1
    for i in s[::2]:
        print(i, end="")

    print()
    # Method 2
    for i in range(len(s)):
        if i % 2 == 0:
            print(s[i], end="")


s = str(input())
utility(s)
```

```python
# Numbers in decreasing order from n
n=int(input())

while n>=0:
    print(n,end=" ")
    n-=1
```

```python
# print the numbers from 1 to x in the order as 12, 22, 32, 42, 52, ... (in increasing order).
x=int(input())
i=1

while i**2<=x:
    print(i**2,end=" ")
    i+=1
```

```python
# print the numbers from 1 to x in the order as 12, 22, 32, 42, 52, ... (in increasing order).
n=int(input())


def positive(n):
    while n>0 :
        print(n-1,end=" ")
        n-=1
        

def negative(n):
    while n<=0 :
        print(n,end=" ")
        n+=1

if  n>0 :
    positive(n)
else : 
    negative(n)
```
