```python
a=input("Enter a value of a : ")
b=input("Enter a valoue of b :")
separator = input()[0] # [0] → Accesses the first character of that string.
```

```python
# Print a and b in a single line, separated by a space, followed by a newline.
print(f"{a} {b}")
#Print a and b separated by a space, but do not end the output with a newline.
print(a,b,sep=" ",end="")
#Print a and b separated by a specified separator, followed by a newline.
print(a,b,sep=separator)
#Print a and b in a single line, with no spaces between them, followed by a newline.
print(a,b,sep="")
```

```python
# print a, n times in a single line without space between them
n=int(input())
print(a*n,sep="")
```

```python
# print a a times and b b times  in a single line separated by c.

a=input()
b=input()
c=input()[0]

print(a*int(a),b*int(b),sep=c)
```
