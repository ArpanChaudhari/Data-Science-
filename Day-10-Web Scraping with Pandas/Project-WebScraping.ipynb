### 2. Read csv from Multiple URLs with pandas

Link : root + season + league

```python
# link structure 
"https://www.football-data.co.uk/mmz4281/" + "2526" + "/" + "E1" + ".csv"
```

```python
# create a root variable
root = "https://www.football-data.co.uk/mmz4281/"
```

### 2.1 Multiple leagues

* https://www.football-data.co.uk/mmz4281/2526/E0.csv
* https://www.football-data.co.uk/mmz4281/2526/E1.csv
* https://www.football-data.co.uk/mmz4281/2526/E2.csv
* https://www.football-data.co.uk/mmz4281/2526/E3.csv
* https://www.football-data.co.uk/mmz4281/2526/Ec.csv


```python
# creating list of leagues
leagues = [
    "E0",
    "E1",
    "E2",
    "E3",
    "Ec",
]

frames = []

# looping through leagues, read multiple csv and append it into a list
for league in leagues:
    df = pd.read_csv(root + "2526" + "/" + league + ".csv")
    frames.append(df)
```

```python
# length of frames
len(frames)
```

### 2.2 Multiple seasons

```python
for season in range(15,26):
    print(str(season)+str(season+1))
```

```python
leagues = [
    "E0",
    "E1",
    "E2",
    "E3",
    "Ec",
]

frames = []

# loops through range of season and insert new column
for league in leagues:
    for season in range(20, 26):
        df = pd.read_csv(root + str(season) + str(season + 1) + "/" + league + ".csv", encoding='unicode_escape')
        df.insert(1,"season",season)
        frames.append(df)
```

```python
# total frames ( 6 season X 5 leagues)
len(frames)
```

### 3. Organize All the Data in dictionary

```python
# create a dictionary with original name of league as key
dict_countries = {
              'Spanish La Liga':'SP1', 'Spanish Segunda Division':'SP2',
              'German Bundesliga':'D1', 
              'English Premier League':'E0', 'English League 1':'E2', 'English League 2':'E3',
             }
```

```python
# get a dictionaty element
dict_countries['Spanish La Liga']
```

```python
for league in dict_countries:
    print(f"{league}:{dict_countries[league]}")
```

```python
dict_historical_data = {}


# looping through key elements 
# to get all season data of perticular league in one dataframe
for league in dict_countries:
    frames=[] # inside loop b's we need to clear older league data to assign new league data
    for season in range(20,26):
        df = pd.read_csv(root + str(season) + str(season + 1) + "/" + dict_countries[league] + ".csv", encoding='unicode_escape')
        df.insert(1,"season",season)
        frames.append(df)
    df_concat= pd.concat(frames)
    dict_historical_data[league] = df_concat
```
