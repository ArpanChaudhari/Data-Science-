```python
import pandas as pd
```

```python
# read and show dataset
df=pd.read_csv("population_total.csv")
df
```

```python
# remove duplicate value
df.dropna(inplace=True)
```

```python
df.pivot(index="year",columns="country",values="population")
```

```python
df_pivot =df.pivot(index="year",columns="country",values="population")
```

```python
df_pivot=df_pivot[['United States', 'India', 'China', 
                          'Indonesia', 'Brazil']]
```

```python
df_pivot
```

# Lineplot

```python
df_pivot.plot(kind="line",xlabel="Year",ylabel="Population",title="Population (1955-2020)",figsize=(10,4))
```

# Barplot

```python
#selecting only one year
df_pivot_2020=df_pivot[df_pivot.index.isin([2020])]
```

```python
df_pivot_2020
```

```python
# transpose dataframe from rows to column or viceversa
df_pivot_2020=df_pivot_2020.T
```

```python
df_pivot_2020
```

```python
df_pivot_2020.plot(kind="bar",color="orange",xlabel="Year",ylabel="Population",title="Population 2020",figsize=(10,4))
```

# Barplot grouped by “n” variables

```python
df_pivot_sample=df_pivot[df_pivot.index.isin([1980,1990,2000,2010])]
```

```python
df_pivot_sample
```

```python
df_pivot_sample.plot(kind="bar",xlabel="Year",ylabel="Population",title="Population",figsize=(10,4))
```

# Piechart

```python
df_pivot_2020.plot(kind="pie",y=2020.0,figsize=(10,4))
```
