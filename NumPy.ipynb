```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

v = np.concat((a, b), axis=0)
v
```

```python
h = np.concat((a, b), axis=1)
h
```

For concatenation to work, the arrays must have the exact same size along every axis except the one you are joining.
* If you use axis=0, the arrays must have matching numbers of columns.
* If you use axis=1, the arrays must have matching numbers of rows.

If they don't match, Python will throw a ValueError.

### broadcasting

```python
a = np.array([[2],[3],[4]])

b = np.array([1,2,3,4])

a+b
```

NumPy compares array dimensions from right to left. Two dimensions are compatible when they are equal or one of them is 1.

(3, 1) 
        -> (3, 4)
(1, 4)

Equal dimensions or One dimension is 1 → compatible
Otherwise → broadcasting error

```python
a = np.array([[3, 4, 5, 6],
       [4, 5, 6, 7],
       [5, 6, 7, 8]])

np.where(a > 5) 
# np.where() is used to find positions where a condition is true, or to choose values based on a condition.
```

```python
np.where(a > 5, 1, 0) # vectorized if-else
```
