# 6) If statement

```python
if <condition>:
    <code>
elif <condition>:
    <code>
    ...
else:
    <code>
```

```python
age = 15

if age>=18:
    print("You're a adult")
elif age>=13:
    print("You're a teenager")
else:
    print("You're a kid")
```

```python
# Else Statement with For Loops
num1=[1,3,5,8]
num2=[1,5,7,9]

def contains_even_no(numbers):
    for e in numbers:
        if e % 2 == 0:
            print("contain Even number")
            break
    else:
        print("not contain Even number")

contains_even_no(num1)
contains_even_no(num2)
```
