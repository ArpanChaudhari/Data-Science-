```python
import pandas  as pd
df_laptops = pd.read_csv("laptop_price.csv",encoding='unicode_escape')
df_laptops.head(5)
```

### Creating a Conditional Column from More Than 2 Choices: np.select()

```python
import numpy as np
```

```python
# create an array based on multiple price tier (+2 choices)

conditions = [
    (df_laptops["Price_euros"] > 3000),
    (df_laptops["Price_euros"] <= 3000) & (df_laptops["Price_euros"] > 2000),
    (df_laptops["Price_euros"] <= 2000) & (df_laptops["Price_euros"] > 800),
    (df_laptops["Price_euros"] <= 800),
]

choices = ["TO Expensive", "Expensive", "Affordable", "Cheap"]
```

```python
# set it to new column

df_laptops["Price_tier"] = np.select(conditions, choices, default="Unknown")
```

```python

df_laptops.head(5)
```

```python
df_laptops.value_counts("Price_tier")
```
