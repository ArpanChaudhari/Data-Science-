```python
import pandas  as pd
df_laptops = pd.read_csv("laptop_price.csv",encoding='unicode_escape')
df_laptops.head(5)
```

### isin() : Single Filtering
 only list-like objects are allowed to be passed to isin()

```python
df_laptops["Company"].isin(["Apple","HP"])
```

```python
df_laptops[df_laptops["Company"].isin(["Apple","HP"])].head(5)
```

### isin(): Multiple Filtering

```python
filter1 = df_laptops["Company"].isin(["Apple","Dell"])
filter2 = df_laptops["TypeName"].isin(["Notebook","Ultrabook"])
```

```python
df_laptops[filter1 & filter2]
```

```python
df_laptops[filter1 & filter2].value_counts("Company")
```

```python
df_laptops[filter1 & filter2].value_counts("TypeName")
```
