```python
import pandas as pd
```

```python
df_laptops = pd.read_csv("laptop_price.csv",encoding='unicode_escape')
```

```python
df_laptops.head(5)
```

### 1. Filter fataframe based on a condition

```python
# find which rows have "Apple" in the "Company" column
df_laptops["Company"] == "Apple"
```

```python
# filter dataframe based on condition
df_laptops[df_laptops["Company"] == "Apple"].head(5)
```

```python
# verify which companies are in the "Company" column with value_counts()
df_laptops[df_laptops["Company"] == "Apple"].value_counts("Company")
```
