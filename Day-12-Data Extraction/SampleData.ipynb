```python
import pandas as pd
```

```python
# reads csv file
df = pd.read_csv('players_20.csv')
# set index
df.set_index('short_name', inplace=True)
# select column
df = df[['long_name', 'age', 'dob', 'height_cm', 'weight_kg', 'nationality', 'club']]
```

```python
# show dataframe
df
```

### sample()
Parameter
* n	- Number of random rows to return
* frac	- Fraction (%) of rows to return
* replace	- Allow duplicate selection (True/False)
* random_state	- Fixed seed for same output every time

```python
df.sample(10,random_state=10)
```

```python
df.sample(5,random_state=10)
```

```python
# extract a random 20% sample of the dataframe
df.sample(frac=0.2,random_state=10)
```

```python
# df.sample(20000) ValueError: Cannot take a larger sample than population when 'replace=False'
df.sample(20000,replace=True)
```

```python
# df.sample(frac=2) ValueError: Replace has to be set to `True` when upsampling the population `frac` > 1.
df.sample(frac=2,replace=True)
```
