# Polymorphism
Polymorphism means "many forms".

* Same method behaves differently for different objects.
* The same method, function, or interface can behave differently depending on the object that uses it

### Types of Polymorphism in Python
* Method Overriding (Runtime Polymorphism)
* Operator Overloading
* Duck Typing (Python-specific)

### 1. Method Overriding (Runtime Polymorphism)
A child class provides its own implementation of a parent's method.

```python
class Payment:

    def pay(self, amount):
        pass

class UPI(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


payments = [
    UPI(),
    CreditCard()
]

for p in payments:
    p.pay(1000)
```

### 2. Operator Overloading
Python operators are polymorphic.

The same operator behaves differently.

```python
print(10 + 20)


print("Hello" + " World")
```

Same operator : +

Different behavior:
* Number -> Addition
* String -> Concatenation

### 3. Duck Typing (Python-Specific Polymorphism)
Python doesn't care about class type.

It only cares whether the object can perform the required action.

```python
class Notification:

    def send(self):
        pass

class Email(Notification):

    def send(self):
        print("Email Sent")


class SMS(Notification):

    def send(self):
        print("SMS Sent")


class WhatsApp(Notification):

    def send(self):
        print("WhatsApp Message Sent")

notifications = [
    Email(),
    SMS(),
    WhatsApp()
]

for n in notifications:
    n.send()
```
