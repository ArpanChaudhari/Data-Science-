# Abstarction:-
Abstraction hides internal implementation and shows only essential details.

```python
class Authentication:

    def login(self, email, password):

        encrepted_password = self.__encrepted(password)

        db_user = self.__find_User(email)

        if db_user == encrepted_password:
            self.__create_session()

            return "login Successfull"

    def __encrepted(self, password):
        return "encrepted"

    def __find_User(self, email):
        return "encrepted"

    def __create_session(self):
        print("Session Created")


auth = Authentication()

print(auth.login("arpan@gmail.com", "1234"))
```

# ABC

### ABC stands for Abstract Base Class.
It comes from Python's built-in abc module:
* from abc import ABC, abstractmethod

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self,c):
        self.color=c
    
    def get_Color(self):
        return self.color
    
    @abstractmethod
    def get_Area(self):
        pass

class Square(Shape):
    def __init__(self,c,side):
        super().__init__(c) # This calls the constructor of the parent class (Shape).
        self.side=side

    def get_Area(self):
        return (self.side * self.side)
    
class Circle(Shape):
    def __init__(self, c,D):
        super().__init__(c)
        self.D=D

    def get_Area(self):
        return (3.14 * self.D * self.D)

# s = Shape("Red")
"""
give an error because abstract classes cannot be instantiated directly.
Only concrete subclasses like Square can create objects.
"""

s=Square("red",5)

print(s.get_Color())
print(s.get_Area())


s1=Circle("Blue",4)

print(s1.get_Color())
print(s1.get_Area())
```
