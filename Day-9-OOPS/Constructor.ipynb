# Constructor (Init)
### A constructor is a special method that executes automatically when an object is created.

```python
class Student :
    def __init__(self,name,age):
        self.name = name
        self.age = age

s1=Student("Arpan",21)

print(s1.name)
print(s1.age)
```

### Self Keyword - Represent the current object

```python
class Student:
    def __init__(self, name):
        self.name = name

    def display(self):
        print("Name : ", self.name)


s1 = Student("Arpan")
s2 = Student("Yash")

s1.display()
```

### self.name = name & self.age = age are instance variable 

```python
# What happens if we don't use self?

class Student:
    def __init__(self, name):
        name = name 
        # The variable name exists only inside the constructor.

    def display(self):
        print("Name : ", self.name)
        # After constructor finishes: AttributeError: 'Student' object has no attribute 'name'


s1 = Student("Arpan")
s2 = Student("Yash")

s1.display()
```
