# Methods 
### Methods are functions defined inside a class

```python
class Methods:
    def add(self, a, b):
        return a + b

obj = Methods()

print(obj.add(75, 725))
```

# Class Variable 
### Shared Among all object

```python
class Student:
    college = "Adani University"


s1 = Student()
s2 = Student()

print(s1.college)
print(s2.college)
```

# Types of Methods

### 1. Instance Method 
Works with object data

```python
class Student :
    def __init__(self,name):
        self.name=name
    
    def display(self):
        print(self.name)

s2=Student("Arpan")

s2.display()
```

### 2. Class Method
Uses class variable

```python
class Student:
    college = "Adani University"

    @classmethod
    # It uses cls instead of self.
    def show_college(cls):
        print(cls.college)

Student.show_college()
```

```python
class Student:
    college = "Adani University"

    @classmethod
    # It uses cls instead of self.
    def show_college(cls):
        print(cls.college)

    @classmethod
    def change_College(cls,new_name):
        cls.college =  new_name

Student.show_college()

Student.change_College("Nirma University")
print(Student.college)
```

### 3. Static Method
Static method do not use 
* self
* cls

Decorator:
@staticmethod

The method is related to the class conceptually but does not need object data or class data.

```python
class Math:

    @staticmethod
    def add(a,b):
        return a+b
    
print(Math.add(10,20))
```
