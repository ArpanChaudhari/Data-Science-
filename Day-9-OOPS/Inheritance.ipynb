# Inheritance
Inheritance allows one class to acquire properties and methods of another class.

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

d=Dog()

d.eat()
d.bark()
```

### Types of Inheritance

### Single Inheritance 
One child class inherits from exactly one parent class.

```python
class Parent:
    def feature_one(self):
        print("Feature from Parent")

class Child(Parent):
    def feature_two(self):
        print("Feature from Child")

obj = Child()
obj.feature_one()
```

### Multilevel Inheritance 
A child class inherits from a parent class, which itself is a child of another class.

```python
class Grandparent:
    def gold(self):
        print("Generational wealth")

class Parent(Grandparent):
    def house(self):
        print("Family home")

class Child(Parent):
    def car(self):
        print("Personal car")

obj = Child()
obj.gold()   # Inherited from Grandparent
obj.house()  # Inherited from Parent
```

### Multiple Inheritance 
One child class inherits from more than one parent class.

```python
class Father:
    def skills_father(self):
        print("Skills from Father")

class Mother:
    def skills_mother(self):
        print("Skills from Mother")

class Child(Father, Mother):
    pass

obj = Child()
obj.skills_father()  # From Father
obj.skills_mother()  # From Mother
```

### Hierarchical Inheritance 
Multiple child classes inherit from a single parent class.

```python
class Vehicle:
    def general_info(self):
        print("Transport vehicle")

class Car(Vehicle):
    pass

class Truck(Vehicle):
    pass

car_obj = Car()
truck_obj = Truck()
car_obj.general_info()   # Shared method
truck_obj.general_info() # Shared method
```

### Hybrid Inheritance 
A combination of two or more of the inheritance types listed above.

```python
class Base:
    def action(self):
        print("Base action")

class ChildA(Base):
    pass

class ChildB(Base):
    pass

# Combines Hierarchical and Multiple inheritance
class GrandChild(ChildA, ChildB):
    pass

print(GrandChild.__mro__) # Displays lookup sequence
```
