```python
arr = [9, 8, 7, 4, 4, 2, 1, 1, 9, 8]
x = 1

def setInsert(arr):
    s=set()
    for i in arr :
        s.add(i)
    return s

def setDisplay(s):
    for i in sorted(s):
        print(i,end=" ")
    print()


def setErase(s,x):
    if x in s:
        s.remove(x)
        print(f"erased {x}")
        setDisplay(s)
    else :
        print(f"{x} not found")

s=setInsert(arr)
setDisplay(s)
setErase(s,x)
```

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 2, 3}

def common_in_set(a,b):
    return (a & b)

def diff(a,b):
    return ( a - b)

def all_in_set(a,b):
    return (a | b)

print(common_in_set(a,b))
print(diff(a,b))
print(all_in_set(a,b))
```
