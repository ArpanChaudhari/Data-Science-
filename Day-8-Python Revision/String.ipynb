```python
# 2. Translate
print(str.translate(str.maketrans("a","A")))
"""
The translate() method is used to replace multiple characters in a string at once using a translation table.
It usually works together with str.maketrans().

"""
s="apple banana"
table=s.maketrans("ab","xy")
print(s.translate(table))

# Remove all vowels.
s="Arpan Chaudhari"
table=s.maketrans("","","aeiouAEIOU")
print(str.translate(table))
```

### maketrans() can take 3 arguments:
    str.maketrans(x, y, z)
    x → characters to replace
    y → replacement characters
    z → characters to delete


### Splitting and Joining

```python
# 1. Split -> List[]
print(str.split())
s="A,B,C"
print(s.split(","))

S="A-B-C-D"
print(S.rsplit("-",2))   # string.rsplit(separator, maxsplit)

# 2. SplitLines -> List[]
name="Arpan\nChaudhari"
print(name.splitlines())

# 3. Join -> String
lst=["Arpan","Chaudhari"]
print("-".join(lst))

# 4. Partition -> Tuple()
print(S.partition("-")) # Splits the string into 3 parts.
print(S.rpartition("-"))
```

```python
# Decimal to binary


n = int(input("Enter a Decimal Number:"))
# in built method
print(bin(n)[2:])

# custom code
binary = ""
while n > 0:
    rem = n % 2
    binary = str(rem) + binary
    n = n // 2


print(n,end="->")
print(binary)
```

```python
# Binary to decimal

n=input("Enter a Binary Number:")

# in built code
print(int(n,2))

# custom code
# M-1
l=len(n)
dec=0
for i in range(0,l):
    rem=int(n)%10
    dec+=rem*(2**i)
    n=int(n)//10


print(n,end="->")
print(dec)
```

```python
n=input("Enter a Binary Number:")
# M-2
dec=0
power=0
for i in n[::-1]:
    dec+= int(i)*(2**power)
    power+=1


print(n,end="->")
print(dec)
```
