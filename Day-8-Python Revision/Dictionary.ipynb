```python
# Longest Subarray with Sum K

arr = [1, 2, 3, 1, 1, 1]
k = 6

prefix_sum = 0
max_len = 0
mp = {}

for i in range(len(arr)):
    prefix_sum += arr[i]

    if prefix_sum == k:
        max_len = i + 1

    if prefix_sum - k in mp:
        length = i - mp[prefix_sum - k]
        max_len = max(max_len, length)

    if prefix_sum not in mp:
        mp[prefix_sum] = i

print(max_len)

# Current Prefix - Previous Prefix = K

```

```python
# Sort according to an Array
"""
Given two integer arrays a1[] and a2[]. 
Sort the first array a1[] such that all the 
relative positions of the elements in the first array 
are the same as the elements in the second array a2[].
Input: a1[] = [2, 1, 2, 3, 4], a2[] = [2, 1, 2]
Output: [2, 2, 1, 3, 4]
"""

def relativeSort(a1,a2):
    mp={}

    # 1. Build frequency map of a1
    for i in a1:
        if i in mp :
            mp[i]+=1
        if i not in mp :
            mp[i]=1

    lst=[]
    
    # 2. Traverse a2
    for i in a2 :

         # Ignore duplicate occurrences in a2
        if i not in mp:
            continue

        x=mp[i]

        # place each element according to its frequency
        for j in range(x):
            lst.append(i)

        # remove it from map
        del mp[i]

    # 3. Collect remaining elements from map
    remaining=[]

    for i in mp:
        x=mp[i]

        for j in range(x):
            remaining.append(i)
            
    # # Sort remaining elements      
    remaining.sort()
    for i in remaining : 
        lst.append(i)

    return lst

a1 = [2,1,2,3,4]
a2 = [2,1,2]

print(relativeSort(a1,a2))
```

```python
# Better Version of above

def relativeSort(a1, a2):
    mp = {}

    # Frequency map
    for num in a1:
        mp[num] = mp.get(num, 0) + 1

    ans = []

    # Elements according to a2
    for num in a2:
        if num in mp:
            ans.extend([num] * mp[num])
            del mp[num]

    # Remaining elements
    remaining = sorted(mp.keys())

    for num in remaining:
        ans.extend([num] * mp[num])

    return ans

a1 = [2,1,2,3,4]
a2 = [2,1,2]

print(relativeSort(a1,a2))
```

```python
# First Repeating Element
"""
Given an array arr[],
find the first repeating element index.
The element should occur more than once and
the index of its first occurrence should be the smallest.

Note:- The position you return should be according to 1-based indexing.
"""


def firstRepeated(arr):
    mp = {}

    for i in arr:
        mp[i] = mp.get(i, 0) + 1
    print(mp)

    rep=[i for i in mp if mp.get(i) > 1 ]
    print(rep)

    if len(rep) == 0 :
            return -1

    min_index = len(arr)
    # x=len(arr)
    for i in rep :
        x=arr.index(i)
        min_index=min(min_index,x)

    
    return min_index+1


a1 = [3,3,4]


print(firstRepeated(a1))
```

```python
# better version
def firstRepeated(arr):
    mp = {}

    # Count frequency
    for num in arr:
        mp[num] = mp.get(num, 0) + 1

    # Find first repeating element
    for i in range(len(arr)):
        if mp[arr[i]] > 1:
            return i + 1  # 1-based indexing

    return -1


a1 = [3, 3, 4]


print(firstRepeated(a1))
```
